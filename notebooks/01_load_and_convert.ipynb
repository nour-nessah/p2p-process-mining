# 01 – Load the XES log and convert it to Parquet

The BPI Challenge 2019 log is a 695 MB XES file. It is read once with PM4Py and saved as Parquet (about 14 MB), which DuckDB can query quickly in the next notebooks.

The raw file `data/BPI_Challenge_2019.xes` and the Parquet file are not tracked in Git (see `.gitignore`).

## 1. Read the XES file and check its size

```python
import pm4py
import pandas as pd
from pathlib import Path

XES_PATH = Path("../data/BPI_Challenge_2019.xes")

df = pm4py.read_xes(str(XES_PATH))
print(df.shape)
print(df.dtypes)
case_col = "case:concept:name"
act_col = "concept:name"
time_col = "time:timestamp"

print("Events:", len(df))
print("Cases:", df[case_col].nunique())
print("Distinct activities:", df[act_col].nunique())
print("Vendors:", df["case:Vendor"].nunique())
print("Period:", df[time_col].min(), "->", df[time_col].max())
print()
print(df[act_col].value_counts())


```

## 2. Save as Parquet
No cleaning here: the data is converted unchanged. Data-quality problems (wrong dates) are found and handled in notebook 02.

```python
OUT_PATH = Path("../data/bpi2019.parquet")
df.to_parquet(OUT_PATH, index=False)
print("Saved:", OUT_PATH, round(OUT_PATH.stat().st_size / 1e6, 1), "MB")
```
